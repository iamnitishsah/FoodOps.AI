```python
df=train_data.merge(center_info, how='left', on='center_id')
```

```python
df.head()
```

```python
df=df.merge(meal_info, how='left', on='meal_id')
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
meal_count=(df.groupby(['week', 'center_id'])['meal_id'].nunique().reset_index(name='unique_meal_count'))
```

```python
meal_count.head()
```

```python
df['num_orders'].describe()
```

```python
df['num_orders'].plot(kind='hist', bins=100, logy=True, figsize=(10, 5), title='Distribution of Number of Orders')
```

The num_orders distribution is highly right-skewed.
Most observations have relatively low order counts, while a small number of observations have very large order counts, producing a long right tail with a few extreme values.

```python
df['discount_percentage']=((df['base_price']-df['checkout_price'])/df['base_price'])*100
```

```python
df['discount_percentage'].plot(kind='hist', bins=100, logy=True, figsize=(10, 5), title='Distribution of Discount Percentage')
```

```python
df.plot(
    kind='scatter',
    x='discount_percentage',
    y='num_orders',
    figsize=(10, 5),
    title='Discount Percentage vs Number of Orders',
    alpha=0.3
)
```

```python
df.head()
```

```python
weekly_orders=(df.groupby('week')['num_orders'].sum().reset_index(name='total_orders'))
```

```python
weekly_orders.head()
```

```python
plt.figure(figsize=(15, 5))
plt.plot(weekly_orders['week'], weekly_orders['total_orders'], marker='o')
plt.title('Total Orders per Week')
plt.xlabel('Week')
plt.ylabel('Total Orders')
plt.grid()
plt.show()
```

```python
promo_effect=(
    df.groupby(['emailer_for_promotion', 'homepage_featured'])['num_orders']
    .agg(['mean', 'count'])
    .reset_index()
)
```

```python
promo_effect['percetage_change']=(
(promo_effect['mean'] - promo_effect['mean'].values[0]) / promo_effect['mean'].values[0]
) * 100
```

```python
promo_effect.head()
```

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(12, 8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap')
plt.show()
```
