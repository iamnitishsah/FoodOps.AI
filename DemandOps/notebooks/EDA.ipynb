```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
train_data = pd.read_csv('../data/raw/train.csv')
train_data.head()
train_data.info()
```

```python
centre_info = pd.read_csv('../data/raw/fulfilment_center_info.csv')
centre_info.head()
centre_info.info()
```

```python
meal_info = pd.read_csv('../data/raw/meal_info.csv')
meal_info.head()
meal_info.info()
```

```python
df = train_data.merge(centre_info, how='left', on='center_id')
df.head()
df.info()
```

```python
df = df.merge(meal_info, how='left', on='meal_id')
df.head()
df.info()
```

```python
df.to_csv('../data/processed/merged_data.csv', index=False)
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Check for duplicates
df.duplicated(subset=['week', 'center_id', 'meal_id']).sum()
```

```python
discount_percentage = (
    (df['base_price'] - df['checkout_price'])
    / df['base_price']
) * 100

discount_percentage.describe()
```

```python
weekly_orders = (
    df.groupby('week')['num_orders']
      .sum()
      .reset_index()
)

weekly_orders.head()
```

```python
plt.figure(figsize=(14, 5))

plt.plot(
    weekly_orders['week'],
    weekly_orders['num_orders'],
    marker='o'
)

plt.xlabel('Week')
plt.ylabel('Total Orders')
plt.title('Total Weekly Orders Across All Centers and Meals')
plt.grid(alpha=0.3)

plt.show()
```

```python
pairs = df[['center_id', 'meal_id']].drop_duplicates()
sample_pairs = pairs.sample(6, random_state=42)
print(sample_pairs)
```

```python
fig, axes = plt.subplots(
    3, 2,
    figsize=(15, 12)
)

axes = axes.flatten()

for ax, (_, row) in zip(axes, sample_pairs.iterrows()):

    center = row['center_id']
    meal = row['meal_id']

    pair = df[
        (df['center_id'] == center) &
        (df['meal_id'] == meal)
    ].sort_values('week')

    ax.plot(
        pair['week'],
        pair['num_orders'],
        marker='o'
    )

    ax.set_title(
        f'Center {center} — Meal {meal}'
    )

    ax.set_xlabel('Week')
    ax.set_ylabel('Num Orders')
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
promo_effect = (
    df.groupby(['emailer_for_promotion', 'homepage_featured'])
      ['num_orders']
      .agg(['mean', 'count'])
      .reset_index()
)

baseline_mean = promo_effect.loc[
    (promo_effect['emailer_for_promotion'] == 0) &
    (promo_effect['homepage_featured'] == 0),
    'mean'
].iloc[0]

promo_effect['pct_change_vs_baseline'] = (
    (promo_effect['mean'] - baseline_mean) / baseline_mean * 100
)

promo_effect
```

```python
category_orders = (
    df.groupby('category')['num_orders']
      .agg(['mean', 'median', 'count', 'sum'])
      .sort_values('mean', ascending=False)
)

category_orders['mean'].plot(
    kind='bar',
    figsize=(10, 5),
    title='Average Orders by Meal Category',
    ylabel='Average Number of Orders',
    xlabel='Meal Category'
)
```

```python
cuisine_orders = (
    df.groupby('cuisine')['num_orders']
      .agg(['mean', 'median', 'count', 'sum'])
      .sort_values('mean', ascending=False)
)

cuisine_orders['mean'].plot(
    kind='bar',
    figsize=(8, 5),
    title='Average Orders by Cuisine',
    ylabel='Average Number of Orders',
    xlabel='Cuisine'
)
```

```python
center_type_orders = (
    df.groupby('center_type')['num_orders']
      .agg(['mean', 'median', 'count', 'sum'])
      .sort_values('mean', ascending=False)
)

center_type_orders['mean'].plot(
    kind='bar',
    figsize=(8, 5),
    title='Average Orders by Center Type',
    ylabel='Average Number of Orders',
    xlabel='Center Type'
)
```
